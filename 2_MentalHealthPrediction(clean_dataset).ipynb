```python
# ==========================================================
# 1. MENGHUBUNGKAN GOOGLE DRIVE
# ==========================================================
from google.colab import drive
import pandas as pd
import os

# Mount Google Drive
drive.mount('/content/drive')

data_path = '/content/drive/MyDrive/03 Dataset/data.csv'
```

```python
# ==========================================================
# 2. MEMBACA DATASET
# ==========================================================
# Mencoba beberapa encoding umum agar file terbaca dengan benar
encodings = ['utf-8', 'latin1', 'utf-16', 'ISO-8859-1']
for enc in encodings:
    try:
        df = pd.read_csv(data_path, encoding=enc)
        print(f"✅ File berhasil dibaca dengan encoding: {enc}")
        break
    except Exception as e:
        last_error = e

print("Ukuran dataset:", df.shape)
df.head()
```

```python
# Cek jumlah missing values pada setiap kolom
missing_values = df.isnull().sum()

# Tampilkan hanya kolom yang memiliki nilai kosong
missing_values[missing_values > 0]
```

```python
#==================================
# MENGHAPUS BARIS YANG TIDAK MEMILIKI TARGET 'DEP'

# Hapus baris yang nilai kolom 'ToDep' kosong
df = df.dropna(subset=['ToDep'])
```

```python
#MENGISI NILAI KOSONG DENGAN MODUL
df['Internet'] = df['Internet'].fillna(df['Internet'].mean())
```

```python
# Cek jumlah missing values pada setiap kolom (tahap 2)
missing_values = df.isnull().sum()

# Tampilkan hanya kolom yang memiliki nilai kosong (tahap 2)
missing_values[missing_values > 0]
```

```python
# ==========================================================
# 4. FEATURE SELECTION (PILIH ATRIBUT RELEVAN)
# ==========================================================
relevant_columns = [
   'Academic','Suicide', 'ToSC', 'APD', 'AHome', 'APH', 'Afear', 'ACS', 'AGuilt', 'AMiscell', 'Partner', 'Friends', 'Parents', 'Relative', 'Profess', 'Phone', 'Doctor', 'Reli', 'Alone', 'Others', 'ToDep', 'Dep'
]

# Pastikan hanya kolom yang ada di dataset yang diambil
selected_columns = [col for col in relevant_columns if col in df.columns]

df_selected = df[selected_columns]

print("\n📊 Jumlah kolom setelah seleksi:", len(df_selected.columns))
print("Kolom yang dipertahankan:")
print(df_selected.columns.tolist())
```

```python
# ==========================================================
# 5. SIMPAN DATASET HASIL SELEKSI KE GOOGLE DRIVE
# ==========================================================
output_path = '/content/drive/MyDrive/03 Dataset/mental_health_selected6.csv'

# Create the directory if it doesn't exist
output_dir = os.path.dirname(output_path)
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

df_selected.to_csv(output_path, index=False)

print(f"\n💾 Dataset hasil seleksi disimpan di Google Drive sebagai:\n{output_path}")
```
